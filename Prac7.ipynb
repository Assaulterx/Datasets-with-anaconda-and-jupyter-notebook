```python
import nltk
nltk.download("all")
```

```python
from nltk import word_tokenize, sent_tokenize
```

```python
corpus = "Max Verstappen is the GOAT of F1. He is a winner of 4 consecutive GP. He is currently playing F1 from Red Bull Racing."
```

```python
print(word_tokenize(corpus))
print(sent_tokenize(corpus))
```

```python
from nltk import pos_tag
```

```python
tokens = word_tokenize(corpus)
print(pos_tag(tokens))
```

```python
from nltk.corpus import stopwords
stop_words= set(stopwords.words("english"))
stop_words
```

```python
tokens = word_tokenize(corpus)
cleaned_tokens = []
for token in tokens:
  if (token not in stop_words):
    cleaned_tokens.append(token)
print(cleaned_tokens)
```

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
```

```python
stemmed_tokens = []
for token in cleaned_tokens:
  stemmed = stemmer.stem(token)
  stemmed_tokens.append(stemmed)
print(stemmed_tokens)
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
lemmatized_tokens = []
for token in cleaned_tokens:
  lemmatized = lemmatizer.lemmatize(token)
  lemmatized_tokens.append(lemmatized)
print(lemmatized_tokens)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from nltk.probability import FreqDist
```

```python
corpus = [
     "Max Verstappen is the GOAT of F1",
     "He is a winner of 4 consecutive GP",
     "He is currently playing F1 from Red Bull Racing"]
```

```python
vectorizer = TfidfVectorizer()
```

```python
vectorizer.fit(tokens)
fdist=FreqDist(tokens)
for word,frequency in fdist.items():
    print(f"{word}:{frequency}")
```

```python
tfidf_matrix = vectorizer.transform(corpus)
print(tfidf_matrix)
```

```python
print(vectorizer.get_feature_names_out())
```
